```python
# create a NumPy array from a list
import numpy as np

arr = np.array([1, 2, 3, 4])
print(arr)
```

```python
# create a NumPy array from a tuple
t = (1,2,3,4)
print(np.array(t))
```

```python
# To find the dimension of array
a = [[3,4],[4,6]],[[3,5],[6,7]]
print(np.ndim(a))
```

```python
#To find the shape of array
a = [[3,4]],[[5,6]]
print(np.shape(a))
```

```python
# convert array to list
arr = np.asarray([3, 4])
lst = arr.tolist()
print(arr)
print(lst)
```
