```python
# 1. Attendance Percentage Calculator
total_classes = 45
attended_class = int(input("Enter the no. of classes attended"))
percentage = (attended_class/total_classes)*100
print(percentage)
if percentage >= 75:
    print("Eligible for exam")
else:
    print("You are not eligible")
```

```python
# 2. Daily expenses
n = int(input("enter the no. of expenses today"))
expense = []
for i in range(n):
    amount = float(input("Enter expense"))
    expense.append(amount)
print(expense)
print("Total expense", sum(expense))
```

```python
# 3. Password strength checker
password = input("Enter the password")
has_upper = False
has_lower = False
has_digit = False
print(password)

if len(password) >= 8:
    for ch in password:
        if ch.isupper():
            has_upper = True
        elif ch.islower():
            has_lower = True
        elif ch.isdigit():
            has_digit = True
    if has_upper and has_lower and has_digit:
        print("Strong password")
    else:
        print("Moderate password")
else:
    print("Weak password")
```

```python
# 4.Simple login validation
correct_user = "admin"
correct_pass = "1234"

user = input("Username: ")
pwd = input("Password: ")
if user == correct_user and pwd == correct_pass:
    print("Login successful")
else :
    print("Invalid credits")
```

```python
#5.Basic To-Do list
tasks = []
while True:
    task = input("Enter the work to do(or exit): ")
    if task == "exit":
        break
    tasks.append(task)
print("\n Your To_Do List:")
for t in tasks:
    print("-",t)
```
