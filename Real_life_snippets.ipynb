```python
# 1. Attendance Percentage Calculator
total_classes = 45
attended_class = int(input("Enter the no. of classes attended"))
percentage = (attended_class/total_classes)*100
print(percentage)
if percentage >= 75:
    print("Eligible for exam")
else:
    print("You are not eligible")
```

```python
# 2. Daily expenses
n = int(input("enter the no. of expenses today"))
expense = []
for i in range(n):
    amount = float(input("Enter expense"))
    expense.append(amount)
print(expense)
print("Total expense", sum(expense))
```
