```python
#read
f = open("data.txt","r")
print(f.read())
f.close()
```

```python
#write
f = open("data.txt", "w")
f.write("I completed my high school in Mount Zion")
f.close()

f = open("data.txt","r")
print(f.read())
```

```python
#append
f = open("data.txt","a")
f.write("\nMy native place is Pudukkottai")
f.close()

f = open("data.txt","r")
print(f.read())
```
