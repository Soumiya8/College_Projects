```python
#To Find the largest and smallest number in a list
lst1 = [4,7,6,1]
largest = max(lst1)
smallest = min(lst1)
print(f"Largest of the list {largest}")
print(f"Smallest of the list {smallest}")
```

```python
#To find the count of elements in list
lst = [3,3,3,5,7,6,5,7,0]
print(lst.count(3))
```

```python
#To remove duplicate elements from list
lst = [1,5,3,2,5,1]
s = set(lst)
l = list(s)
print(l)
```

```python
#find the sum of tuple
t = (1,32,5)
a = sum(t)
print(a)
```

```python
#To find the index of a given element in a tuple
t = (1,4,6)
n = int(input("Enter the Element : "))
print(t.index(n))
```
