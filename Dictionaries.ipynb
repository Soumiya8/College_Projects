```python
#dictionry basic syntax
cartoons = { "Doraemon": "Creativity" , 
           "Shin Chan" : "Happy",
           "tom and Jerry" : "Bond",
           "Chotta Bheem": "Laddu",
           "Dora" : "Ultra Blind",
           "Jackie Chan" : "Fight"}
print(cartoons.items())
#key value
print(cartoons["Doraemon"])
#get() safe
print(cartoons.get("Ninja Hattori"))
#add new key value
cartoons["Stuart Little"] = "Empathy"
print(cartoons)
#deleting elements
del cartoons["Jackie Chan"]
print(cartoons)
#other method to delete
print(cartoons.pop("Dora"))
#to find len
len(cartoons)
#Key names
cartoons.keys()
cartoons.values()
```

```python
#Looping in dict
for i in cartoons:
    print(i, end =",")
print()
print()
for ch in cartoons.values():
    print(ch, end=",")

for character,characteristic in cartoons.items():
    print(f"{character},{characteristic}")
```

```python
#nested Dict
food = {"fruits": {"apple":"red",
                   "banana":"yellow",
                   "orange":"orange"},
        "snacks": {"lays":"green",
                   "lottie pi":"brown"}}

print(food["fruits"]["apple"])

print(food["snacks"].values())

#using loop
for ch in food["snacks"].values():
    print(ch)
```
