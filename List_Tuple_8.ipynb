```python
# perform linear search. The function accepts a 
#tuple of ‘n’ elements and a search element.. The function should return the 
#index position of the search element
def linear_search(t,key):
    for i in range(len(t)):
        if t[i] == key:
            print(i)
            break
    return -1
t = (3,4,6,2,0)
print(linear_search(t,3))
```

```python
#function ‘count_data_tuple’ that takes a tuple of numbers 
#and returns the nested tuple of the form (data, count).
def count_data_tuples(t):
    result = []
    for i in t:
        result.append((i,t.count(i)))
    return tuple(result)
t = (2,4,5,2,4,6,7,4)
print(count_data_tuples(t))
```
