```python
#1.Dice Roll game
import random
dice_C = [1,2,3,4,5,6]
dice_H = [1,2,3,4,5,6]
print("Computer:")
computer = random.choice(dice_C)
print(computer)
print("Player:")
human = random.choice(dice_H)
print(human)
if human > computer  :
    print("You won")
else:
    print("You Lose")
```

```python
#2.Number guessing game
import random
while True:
    number = random.randint(1,10)
    guess = int(input("Guess a number between 1 and 10 or 0 to exit :"))
    if guess == 0:
        break
    elif number == guess:
        print("Correct! You guessed it")
        break
    else:
        print("Wrong! The number was",number)
```
